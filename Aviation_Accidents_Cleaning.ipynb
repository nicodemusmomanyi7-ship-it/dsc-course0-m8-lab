## Data Cleaning

### Filtering aircrafts and events

We want to filter the dataset to include aircraft that the client is interested in an analysis of:
- inspect relevant columns
- figure out any reasonable imputations
- filter the dataset

```python
# checking accidents vs incidents
print(df["Investigation.Type"].value_counts(dropna=False))
df = df[df["Investigation.Type"] == "Accident"]
print("After keeping accidents:", df.shape)
```

```python
#filtering by year
df["Event.Date"] = pd.to_datetime(df["Event.Date"])
df = df[df["Event.Date"] >= "1983-01-01"]
print(df.shape)
```

```python
#removing amateeur built
df["Amateur.Built"] = df["Amateur.Built"].fillna('No')
df = df[df["Amateur.Built"] == 'No']
print(df.shape)
```

```python

## assume the 'Unknown' and 'UNK' as missing
df['Aircraft.Category'] = df['Aircraft.Category'].replace({'Unknown': np.nan, 'UNK': np.nan})

# Make a temporary key for each plane type, e.g. 'CESSNA 172'
df['key'] = df['Make'].str.upper().str.strip() + ' ' + df['Model'].str.upper().str.strip()

# Finding the most common known category for each plane type
lookup = df.dropna(subset=['Aircraft.Category']).groupby('key')['Aircraft.Category'].agg(lambda s: s.mode()[0])

## Fill the missing categories from that lookup, then remove the temporary key
df['Aircraft.Category'] = df['Aircraft.Category'].fillna(df['key'].map(lookup))
df = df.drop(columns='key')

print('Still missing:', df['Aircraft.Category'].isna().sum())
```

```python
## keeping the Airlines
df = df[df['Aircraft.Category'] == 'Airplane']
print(df.shape)
```

### Cleaning and constructing Key Measurables

Injuries and robustness to destruction are a key interest point for the client. Clean and impute relevant columns and then create derived fields that best quantifies what the client wishes to track. **Use commenting or markdown to explain any cleaning assumptions as well as any derived columns you create.**

**Construct metric for fatal/serious injuries**

*Hint:* Estimate the total number of passengers on each flight. The likelihood of serious / fatal injury can be estimated as a fraction from this.

```python

## The four injury columns together that covered everyone on board
inj_cols = ['Total.Fatal.Injuries', 'Total.Serious.Injuries', 'Total.Minor.Injuries', 'Total.Uninjured']

## Dropping rows where all four are empty  
df = df[df[inj_cols].notna().any(axis=1)]

## Filling the remaining empty spots with 0
df[inj_cols] = df[inj_cols].fillna(0)

##  The setimated number of people on board
df['Total.Onboard'] = df[inj_cols].sum(axis=1)
df = df[df['Total.Onboard'] > 0]

print(df.shape)
```

```python
df['Fatal.Serious.Count'] = df['Total.Fatal.Injuries'] + df['Total.Serious.Injuries']

# Fraction of people on board who were killed or seriously hurt (0 to 1)
df['Fatal.Serious.Rate'] = df['Fatal.Serious.Count'] / df['Total.Onboard']

df[['Total.Onboard', 'Fatal.Serious.Count', 'Fatal.Serious.Rate']].describe()
```

**Aircraft.Damage**
- identify and execute any cleaning tasks
- construct a derived column tracking whether an aircraft was destroyed or not.

```python
print(df['Aircraft.damage'].value_counts(dropna=False))
```

```python
# 'Unknown' gives no information, so treat it as missing
df['Aircraft.damage'] = df['Aircraft.damage'].replace('Unknown', np.nan)

# Drop the rows with no damage information
df = df[df['Aircraft.damage'].notna()]

# New column: 1 if the aircraft was destroyed, 0 if not
df['Destroyed'] = (df['Aircraft.damage'] == 'Destroyed').astype(int)

print(df.shape)
print(df['Destroyed'].mean())
```

### Investigate the *Make* column
- Identify cleaning tasks here
- List cleaning tasks clearly in markdown
- Execute the cleaning tasks
- For your analysis, keep Makes with a reasonable number (you can put the threshold at 50 though lower could work as well)

```python
# Drop rows with no make
df = df.dropna(subset=['Make'])
print('Makes before cleaning:', df['Make'].nunique())

# Upper case and remove extra spaces
df['Make'] = df['Make'].str.upper().str.strip()

# Remove punctuation
df['Make'] = df['Make'].str.replace(r"[.,']", '', regex=True)

# Remove company words at the end (repeat 3 times in case a name has more than one)
for _ in range(3):
    df['Make'] = df['Make'].str.replace(r'\s+(AIRCRAFT|INC|CORP|CORPORATION|CO|COMPANY|LTD)$', '', regex=True)

# Merge different spellings of the same make
df['Make'] = df['Make'].replace({'BEECHCRAFT': 'BEECH', 'AIRBUS INDUSTRIE': 'AIRBUS', 'DEHAVILLAND': 'DE HAVILLAND'})

print('Makes after cleaning:', df['Make'].nunique())
```

```python

# Count how many records each make has
make_counts = df['Make'].value_counts()
print(make_counts.head(10))

# Keep only makes with at least 50 records
keep_makes = make_counts[make_counts >= 50].index
df = df[df['Make'].isin(keep_makes)]

print('Makes kept:', len(keep_makes))
print(df.shape)
```

### Inspect Model column
- Get rid of any NaNs.
- Inspect the column and counts for each model/make. Are model labels unique to each make?
- If not, create a derived column that is a unique identifier for a given plane type.

```python

# Dropping  rows with no model
df = df.dropna(subset=['Model'])

# Upper case, strip spaces, and remove spaces and hyphens inside the name
# so '172 N', '172-N' and '172N' count as the same model
df['Model'] = df['Model'].str.upper().str.strip().str.replace(r'[\s\-]+', '', regex=True)

print('Unique models:', df['Model'].nunique())
df['Model'].value_counts().head(10)
```

```python
### Are model names shared across makes?
makes_per_model = df.groupby('Model')['Make'].nunique()
print('Models used by more than one make:', (makes_per_model > 1).sum())
```

```python
## Model alone is not unique, so combine make and model into one plane-type label
df['Make.Model'] = df['Make'] + ' ' + df['Model']

print('Unique plane types:', df['Make.Model'].nunique())
print(df['Make.Model'].value_counts().head(10))
```

### Cleaning other columns
- there are other columns containing data that might be related to the outcome of an accident. We list a few here:
- Engine.Type
- Weather.Condition
- Number.of.Engines
- Purpose.of.flight
- Broad.phase.of.flight

Inspect and identify potential cleaning tasks in each of the above columns. Execute those cleaning tasks. 

**Note**: You do not necessarily need to impute or drop NaNs here.

```python

# filling missing values in a column with the most common value (mode)
# of the same plane type (Make.Model)
def fill_with_mode(col):
    modes = df.groupby('Make.Model')[col].agg(lambda s: s.mode()[0] if s.notna().any() else np.nan)
    return df[col].fillna(df['Make.Model'].map(modes))
```

```python
## Engine.Type
print(df['Engine.Type'].value_counts(dropna=False))

# 'Unknown' and 'UNK' mean the same as empty; 'Geared Turbofan' is the same type as 'Turbo Fan'
df['Engine.Type'] = df['Engine.Type'].replace({'Unknown': np.nan, 'UNK': np.nan, 'Geared Turbofan': 'Turbo Fan'})

# Filling the gaps using the most common engine type for the same plane type
df['Engine.Type'] = fill_with_mode('Engine.Type')
print(df['Engine.Type'].value_counts(dropna=False))
```

```python

## Number.of.Engines
print(df['Number.of.Engines'].value_counts(dropna=False))

# An airplane can't have 0 engines, so treat 0 as missing
df['Number.of.Engines'] = df['Number.of.Engines'].replace(0, np.nan)

# Filling  the gaps using the most frequent for the same plane type
df['Number.of.Engines'] = fill_with_mode('Number.of.Engines')
print(df['Number.of.Engines'].value_counts(dropna=False))
```

```python
## Weather.Condition: only VMC (good visibility) and IMC (poor visibility) are real values
print(df['Weather.Condition'].value_counts(dropna=False))
df['Weather.Condition'] = df['Weather.Condition'].replace({'UNK': np.nan, 'Unk': np.nan})
print(df['Weather.Condition'].value_counts(dropna=False))
```

```python

## Purpose.of.flight
print(df['Purpose.of.flight'].value_counts(dropna=False))

# 'Unknown' is not a real purpose; 'Air Race show' and 'Air Race/show' are the same thing
df['Purpose.of.flight'] = df['Purpose.of.flight'].replace({'Unknown': np.nan, 'Air Race show': 'Air Race/show'})
print(df['Purpose.of.flight'].value_counts(dropna=False))
```

### Column Removal
- inspect the dataframe and drop any columns that have too many NaNs

```python
## Percentage of missing values per column, worst first
missing_pct = df.isna().mean().sort_values(ascending=False) * 100
print(missing_pct.round(1).head(10))

# Drop columns that are more than 50% empty
to_drop = missing_pct[missing_pct > 50].index.tolist()
print('Dropping:', to_drop)
df = df.drop(columns=to_drop)
df.shape
```
