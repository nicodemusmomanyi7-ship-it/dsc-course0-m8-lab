### Cleaning and constructing Key Measurables

Injuries and robustness to destruction are a key interest point for the client. Clean and impute relevant columns and then create derived fields that best quantifies what the client wishes to track. **Use commenting or markdown to explain any cleaning assumptions as well as any derived columns you create.**

```python
#the total number of injuries
AviationData_Copy["Total.Injuries"] = (
    AviationData_Copy["Total.Fatal.Injuries"]
    + AviationData_Copy["Total.Serious.Injuries"]
    + AviationData_Copy["Total.Minor.Injuries"]
    + AviationData_Copy["Total.Uninjured"]
)
AviationData_Copy
```

```python
#estimating the total number of passengers
AviationData_Copy["Total.Passengers"] = (
    AviationData_Copy["Total.Fatal.Injuries"]
    + AviationData_Copy["Total.Serious.Injuries"]
    + AviationData_Copy["Total.Minor.Injuries"]
    + AviationData_Copy["Total.Uninjured"]
)
```

```python
#since the client requires reliable aircraft safety comparisons, I consider excluding records with incomplete injury information from the total injury-rate calculation
AviationData_Copy = AviationData_Copy.dropna(subset="Total.Injuries")
AviationData_Copy.head()
```

**Construct metric for fatal/serious injuries**

*Hint:* Estimate the total number of passengers on each flight. The likelihood of serious / fatal injury can be estimated as a fraction from this.

```python
AviationData_Copy["Fatal_Serious_Injury_Rate"] = (
    AviationData_Copy["Total.Fatal.Injuries"] + AviationData_Copy["Total.Serious.Injuries"]
) / AviationData_Copy["Total.Passengers"]
```

```python
#engine type
AviationData_Copy["Engine.Type"].unique()
```

```python
AviationData_Copy["Engine.Type"].isna().sum()
```

```python
columns_to_check = [
    "Engine.Type",
    "Weather.Condition",
    "Number.of.Engines",
    "Purpose.of.flight",
    "Broad.phase.of.flight"
]

for col in columns_to_check:
    print(f"\n--- {col} ---")
    print(AviationData_Copy[col].value_counts(dropna=False))
```

```python
AviationData_Copy["Engine.Type"] = (
    AviationData_Copy["Engine.Type"]
    .str.strip()
    .str.upper()
)
```

```python
AviationData_Copy["Weather.Condition"].value_counts(dropna=False)
```

```python
AviationData_Copy["Weather.Condition"] = (
    AviationData_Copy["Weather.Condition"]
    .str.strip()
    .str.upper()
    .replace(["UNK", "UNKNOWN", ""], np.nan)
)
```

```python
print(AviationData_Copy["Number.of.Engines"].value_counts(dropna=False))
print(AviationData_Copy["Number.of.Engines"].dtype)
```

```python
AviationData_Copy["Number.of.Engines"] = pd.to_numeric(
    AviationData_Copy["Number.of.Engines"],
    errors="coerce"
)

print(
    AviationData_Copy.loc[
        AviationData_Copy["Number.of.Engines"].notna()
        & (
            (AviationData_Copy["Number.of.Engines"] < 1)
            | (AviationData_Copy["Number.of.Engines"] % 1 != 0)
        ),
        "Number.of.Engines"
    ].value_counts()
)
```

```python
AviationData_Copy["Purpose.of.flight"].value_counts(dropna=False)
```

```python
AviationData_Copy["Purpose.of.flight"] = (
    AviationData_Copy["Purpose.of.flight"]
    .str.strip()
    .str.upper()
)
```

```python
AviationData_Copy["Broad.phase.of.flight"].value_counts(dropna=False)
```

```python
AviationData_Copy["Broad.phase.of.flight"] = (
    AviationData_Copy["Broad.phase.of.flight"]
    .replace(["UNKNOWN", "UNK", ""], np.nan)
)
```

```python
for col in columns_to_check:
    print(f"\n--- {col} ---")
    print(AviationData_Copy[col].value_counts(dropna=False))
```
